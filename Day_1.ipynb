```python
# there ar efew types of arrays
# 1D array
# 2D array
# multidimensional array

d1_array = np.array([5,6,7,8,9])
d2_array = np.array([[1,2,3,4],[3,4,5,6]]) #important that for a 2d array we need to a add a extra square braket
# and if the first has 4 elements then the next list should also must have 5 elelments or else it will show value error

print(d1_array)
print(d2_array)
```

```python
# now lets see about the numpy properties
#roperty	Meaning
#ndim	    Number of dimensions
#shape	    Number of elements along each dimension
#size	    Total number of elements
#dtype	    Data type of elements

print(d2_array.shape)
print(d2_array.size)
print(d2_array.ndim)
print(d2_array.dtype)
```

```python
a = np.array([2, 4, 6, 8])

print(a + 2)
print(a * 3)
print(a - 1)
print(a / 2)
```

```python
marks = np.array([50, 60, 70, 80, 90])

print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.sum(marks))
```

```python
import numpy as np

x = np.array([1, 2, 6, 2, 8, 9])

# Median is built right into NumPy
print("Median:", np.median(x))

# For mode, SciPy is typically used alongside NumPy:
from scipy import stats
print("Mode:", stats.mode(x))
```
