```python
import numpy as np
import pandas as pd

# A Series is a one-dimensional labelled data structure.
marks = pd.Series([45,65,76,86])
print(marks)

# Series  "S" should be in capital letter
```

```python
print(marks[0])
print(marks[2])
```

```python
# DataFrame
# A DataFrame is a 2 dimensional array consisting of rows and columns

marks = pd.DataFrame({"age":[34,45,43,65,35],"marks":[34,98,68,94,24],"day":["monday","tuesday","wednesday","thursday","friday"]})
print(marks)
```

```python
#exploring a dataset using pandas

print(marks.head())  #first five rows
```

```python
print(marks.tail()) #last 5 rows
```

```python
print(marks.shape)
```

```python
print(marks.columns)
```

```python
print(marks.info()) #gives the information about the dataset
```

```python
print(marks.describe()) #shows statis tical data abouot the numerical columns
```

```python
print(marks["marks"])
```

```python
print(marks[["age" , "marks"]])
```

```python
high_marks = marks[marks["marks"]>60]
print(high_marks)
```

```python
max_age = marks[marks["age"]>=40]
print(max_age)
```

```python
print(marks["marks"].mean())
print(marks["marks"].max())
print(marks["marks"].min())
print(marks["marks"].sum())
```

```python
sorted_marks = marks.sort_values(by="marks", ascending = True)

print(sorted_marks)
```

```python
#handling missing values

data={
    "name":["arun","priya","ram","shyam"],
    "marks":[85,np.nan,95,34]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df_clean=df.dropna()
```

```python
df["marks"] = df["marks"].fillna(df["marks"].mean())
print(df)
```
